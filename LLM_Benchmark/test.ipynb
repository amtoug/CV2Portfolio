```python
def prompt(cv_text):
  return f'''
You are a professional CV information extraction system.

Your task is to extract structured information from the CV provided below and return it as a single valid JSON object.

STRICT RULES:
1. Return ONLY valid JSON.
2. Do NOT return Markdown, explanations, comments, or code fences.
3. Do NOT invent, infer, assume, or hallucinate information.
4. Extract information ONLY from the CV.
5. Preserve the original meaning of the CV.
6. If information is missing, use an empty string "" or an empty array [].
7. Use EXACTLY the JSON schema provided below.
8. Do NOT add, remove, or rename any fields.
9. Keep the same structure for every CV.
10. Keep descriptions concise while preserving important technical information.
11. Do not duplicate the same information across multiple sections.
12. Preserve proper names of people, organizations, institutions, technologies, certifications, and projects.
13. Preserve dates and periods as they appear in the CV.
14. Do not convert or normalize dates unless necessary for extraction.
15. Skills must contain individual skills or technologies, not complete sentences.
16. Languages must contain the languages explicitly mentioned in the CV.
17. For experience, extract internships, jobs, freelance experiences, and other professional experiences explicitly mentioned.
18. For projects, extract explicitly identified academic, personal, professional, or portfolio projects.
19. Do not create a project from a simple mention of a technology.
20. If a section does not exist in the CV, return an empty array.

JSON SCHEMA:

{{
  "name": "",
  "title": "",
  "summary": "",
  "skills": [],
  "education": [
    {{
      "degree": "",
      "field": "",
      "institution": "",
      "period": ""
    }}
  ],
  "experience": [
    {{
      "role": "",
      "organization": "",
      "period": "",
      "description": ""
    }}
  ],
  "projects": [
    {{
      "name": "",
      "description": "",
      "technologies": []
    }}
  ],
  "languages": []
}}

CV:
{cv_text}
'''
```

```python
import time
from is_valid import validate_schema, is_valid_json
from openai import RateLimitError

schema_valid_count = []
valid_json_count = []
latencies = []

keys = (GroqKey2, GroqKey3, GroqKey4)


results = []

for model, GroqKey in zip(models, keys):

    valid_json_count = []
    schema_valid_count = []
    latencies = []

    print(f"\n========== {model} ==========")

    for cv in fichiers:

        with open(cv, "r", encoding="utf-8") as f:
            cv_text = f.read()

        for attempt in range(3):

            try:
                start = time.perf_counter()

                response = llm_Groq(model, GroqKey).invoke(
                    prompt(cv_text)
                ).content

                latency = time.perf_counter() - start

                valid_json_count.append(
                    is_valid_json(response)
                )

                schema_valid_count.append(
                    validate_schema(response)
                )

                latencies.append(latency)

                print(
                    f"{cv} | "
                    f"JSON={valid_json_count[-1]} | "
                    f"Schema={schema_valid_count[-1]} | "
                    f"{latency:.2f}s"
                )

                break

            except RateLimitError as e:

                if attempt < 2:
                    print(f"429 - waiting 2 seconds...")
                    time.sleep(2)
                else:
                    print(f"429 - skipped: {cv}")

    total_tests = len(valid_json_count)

    if total_tests > 0:

        json_rate = sum(valid_json_count) / total_tests * 100
        schema_rate = sum(schema_valid_count) / total_tests * 100
        avg_latency = sum(latencies) / total_tests

        results.append({
            "model": model,
            "json_valid_rate": json_rate,
            "schema_valid_rate": schema_rate,
            "average_latency": avg_latency,
            "tests": total_tests
        })

        print(f"\nJSON validity : {json_rate:.2f}%")
        print(f"Schema validity : {schema_rate:.2f}%")
        print(f"Average latency : {avg_latency:.2f}s")
```

```python
import pandas as pd

df=pd.DataFrame(results)
df
```

```python
import matplotlib.pyplot as plt

models_names = [r["model"] for r in results]
json_rates = [r["json_valid_rate"] for r in results]
schema_rates = [r["schema_valid_rate"] for r in results]
latencies = [r["average_latency"] for r in results]
```

```python
fig, axes = plt.subplots(2, 1, figsize=(12, 10))


x = range(len(models_names))
width = 0.35

bars1 = axes[0].bar(
    [i - width / 2 for i in x],
    json_rates,
    width,
    label="JSON Validity"
)

bars2 = axes[0].bar(
    [i + width / 2 for i in x],
    schema_rates,
    width,
    label="Schema Validity"
)

axes[0].set_title(
    "LLM Structured Output Reliability",
    fontsize=15
)

axes[0].set_ylabel("Validity (%)")
axes[0].set_xticks(list(x))
axes[0].set_xticklabels(models_names)
axes[0].set_ylim(0, 110)

axes[0].bar_label(
    bars1,
    fmt="%.1f%%",
    padding=3
)

axes[0].bar_label(
    bars2,
    fmt="%.1f%%",
    padding=3
)

axes[0].legend()


# =====================================================
# 2. Average Latency
# =====================================================

bars3 = axes[1].bar(
    models_names,
    latencies
)

axes[1].set_title(
    "LLM Average Inference Latency",
    fontsize=15
)

axes[1].set_ylabel("Latency (seconds)")
axes[1].set_xticks(range(len(models_names)))
axes[1].set_xticklabels(models_names)

axes[1].bar_label(
    bars3,
    fmt="%.2f s",
    padding=3
)


# =====================================================
# Final layout
# =====================================================

fig.suptitle(
    "LLM Benchmark — CV to Structured JSON",
    fontsize=18,
    fontweight="bold"
)

plt.tight_layout(rect=[0, 0, 1, 0.96])

plt.show()
```
